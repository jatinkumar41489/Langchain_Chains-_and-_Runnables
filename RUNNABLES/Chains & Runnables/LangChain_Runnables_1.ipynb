# How the Components of LangChain were Designed .
 **1) Model (LLM)**  
 **2) Prompt**  
 **3) Chains**

**First Component ---> **Dummy LLM Class****

```python
import random

class NakliLLM:
  #Constructor
  def __init__(self):
    print("LLM Created")

  def predict(self,prompt):

    response_list=[
        "Delhi is the capital of India",
        "ILP is cricket league",
        "AI stands for Artificial Intelligence"
    ]

    return {'response' : random.choice(response_list)}
```

```python
llm=NakliLLM()
```

```python
llm.predict("What is the capital of India") # Response will be random because it is not a actual llm but a dummy llm class
```

**Second Component ---> Prompt Class**

```python
class NakliPromptTemplate:

  # Constructor having two output template and input_variables
  def __init__(self, template, input_variables):
    self.template = template
    self.input_variables = input_variables

  # Format Function in Prompt Class
  def format(self, input_dict):
    return self.template.format(**input_dict) # Unpack the input_dict
```

```python
template = NakliPromptTemplate(
    template="Write a {length} poem about {topic}",
    input_variables=["lenght","topic"]
)
```

```python
prompt = template.format(
    {
        "length":"short",
        "topic":"AI"
    }
)
```

# Now you need to make an appplication by using these two components (LLM and Prompt )

```python
llm = NakliLLM()
```

```python
llm.predict(prompt)
```

**Third Component ---> Chain :**  

Chain will connect both LLM and Prompt

```python
class NakliLLMChain:

  def __init__(self, llm, prompt):
    self.llm = llm
    self.prompt = prompt

  def run(self, input_dict):
    final_prompt = self.prompt.format(input_dict) # This fill automatically trigger this function ---> def format(self, input_dict):
    result = self.llm.predict(final_prompt) # This fill automatically trigger this function ---> def predict(self,prompt):

    return result['response']
```

#LLM APPLICATION :
**So first we need of Prompt , LLM then a chain that will connect both prompt and LLM**

```python
template = NakliPromptTemplate(
    template="Write a {length} poem about {topic}",
    input_variables=["lenght","topic"]
)
```

```python
llm = NakliLLM()
```

```python
chain = NakliLLMChain(llm, template)
```

```python
chain.run(
    {
        "topic":"Delhi",
        "length":"short"
    }
)
```

# But at this point LangChain's team observes that this Chain (class NakliLLMChain:) class is not flexible .

**Reason of failure:**  
1) We Can not make an application for parallel execution of task.
3) More than one LLM Calls are not possible.
2) Two sequential chains is not possible.  
example:
Chain1 (Generate a Joke) ---> Chain2 (Explain that Joke)

**Problem :**  **No Standardization means :**

To interact with **llm** you have method **predict()**.  
To interact with **Prompt** you have method **format()**.  
To interact with **Chains** you have method **run()**.
